```python
# Importer les pipelines
import sys
sys.path.insert(0, '/workspace/projects/freshkart_migration/src')

from pandas_freshkart.pipeline import FreshKartPandasPipeline
from pyspark_freshkart.pipeline import FreshKartPySparkPipeline
from pyspark.sql import SparkSession

# Créer Spark
spark = SparkSession.builder.appName("FreshKart").master("local[*]").getOrCreate()

# Exécuter Pandas
pandas_pipeline = FreshKartPandasPipeline("/workspace/data/march-input")
result_pandas = pandas_pipeline.run_full_pipeline()
print(f"Pandas: {len(result_pandas)} lignes")

# Exécuter PySpark
pyspark_pipeline = FreshKartPySparkPipeline("/workspace/data/march-input", spark)
result_pyspark = pyspark_pipeline.run_full_pipeline()
print(f"PySpark: {result_pyspark.count()} lignes")

result_pandas.head()
```
